```python
from sklearn.datasets import load_breast_cancer
import pandas as pd

# Load the dataset
data = load_breast_cancer()

# Create DataFrame
X = pd.DataFrame(data.data, columns=data.feature_names)
y = pd.Series(data.target, name="target")

# Display first 5 rows
print("First 5 rows of the dataset:")
display(X.head())

# Display dataset shape
print("\nDataset shape:", X.shape)

# Display target values
print("\nTarget values:")
print(y.value_counts())
```

```python
# Check for missing values
print("Missing values in each column:")
print(X.isnull().sum().sum())

# Check for duplicate rows
print("\nNumber of duplicate rows:")
print(X.duplicated().sum())

# Check data types
print("\nData types:")
print(X.dtypes.value_counts())

# Display statistical summary
print("\nStatistical summary:")
display(X.describe())
```

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

# Display the shapes
print("Training feature set shape:", X_train.shape)
print("Testing feature set shape:", X_test.shape)

print("\nTraining target shape:", y_train.shape)
print("Testing target shape:", y_test.shape)

# Check class distribution
print("\nTraining class distribution:")
print(y_train.value_counts())

print("\nTesting class distribution:")
print(y_test.value_counts())
```

```python
from sklearn.preprocessing import StandardScaler

# Create the scaler
scaler = StandardScaler()

# Fit the scaler only on the training data
X_train_scaled = scaler.fit_transform(X_train)

# Use the same scaler to transform the test data
X_test_scaled = scaler.transform(X_test)

# Convert back to DataFrame for easy inspection
X_train_scaled_df = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)

X_test_scaled_df = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns
)

# Display the first 5 rows of scaled training data
print("First 5 rows of scaled training data:")
display(X_train_scaled_df.head())

# Check mean and standard deviation
print("\nMean of scaled features:")
print(X_train_scaled_df.mean().head())

print("\nStandard deviation of scaled features:")
print(X_train_scaled_df.std().head())
```

```python
from sklearn.svm import SVC

# Create Linear Kernel SVM
svm_linear = SVC(
    kernel='linear',
    C=1.0
)

# Train the model
svm_linear.fit(X_train_scaled, y_train)

print("Linear Kernel SVM model trained successfully.")

print("\nModel parameters:")
print("Kernel:", svm_linear.kernel)
print("C:", svm_linear.C)
```

```python
from sklearn.model_selection import GridSearchCV

# Define the parameter grid
param_grid = {
    'C': [0.01, 0.1, 1, 10, 100]
}

# Create a Linear SVM
svm = SVC(kernel='linear')

# Perform 5-fold cross-validation
grid_search = GridSearchCV(
    estimator=svm,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy'
)

# Train and tune the model
grid_search.fit(X_train_scaled, y_train)

# Display results
print("Best C value:", grid_search.best_params_['C'])
print("Best cross-validation accuracy:", grid_search.best_score_)

print("\nBest Linear SVM model:")
print(grid_search.best_estimator_)
```

```python
# Get the best SVM model from GridSearchCV
best_svm = grid_search.best_estimator_

# Make predictions on the test data
y_pred = best_svm.predict(X_test_scaled)

# Display the first 20 predictions
print("First 20 predicted values:")
print(y_pred[:20])

print("\nFirst 20 actual values:")
print(y_test.values[:20])
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

# Calculate evaluation metrics
accuracy = accuracy_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

# Display the results
print("SVM Classification Performance")
print("--------------------------------")
print(f"Accuracy  : {accuracy:.4f}")
print(f"Precision : {precision:.4f}")
print(f"Recall    : {recall:.4f}")
print(f"F1 Score  : {f1:.4f}")
```

```python
from sklearn.metrics import confusion_matrix
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate confusion matrix
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)

# Visualize confusion matrix
plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt='d',
    cmap='Blues',
    xticklabels=['Malignant (0)', 'Benign (1)'],
    yticklabels=['Malignant (0)', 'Benign (1)']
)

plt.xlabel("Predicted Class")
plt.ylabel("Actual Class")
plt.title("Confusion Matrix - Linear SVM")
plt.show()
```

```python
from sklearn.datasets import load_wine
import pandas as pd

# Load the Wine dataset
wine = load_wine()

# Create DataFrame for features
X_wine = pd.DataFrame(
    wine.data,
    columns=wine.feature_names
)

# Create target Series
y_wine = pd.Series(
    wine.target,
    name="target"
)

# Display first 5 rows
print("First 5 rows of the Wine dataset:")
display(X_wine.head())

# Display dataset shape
print("\nDataset shape:", X_wine.shape)

# Display target classes
print("\nTarget classes:")
print(y_wine.value_counts().sort_index())
```

```python
from sklearn.preprocessing import StandardScaler

# Create StandardScaler
scaler_wine = StandardScaler()

# Standardize the Wine dataset
X_wine_scaled = scaler_wine.fit_transform(X_wine)

# Convert scaled data back to DataFrame
X_wine_scaled_df = pd.DataFrame(
    X_wine_scaled,
    columns=X_wine.columns
)

# Display first 5 rows
print("First 5 rows of standardized Wine dataset:")
display(X_wine_scaled_df.head())

# Check mean of standardized features
print("\nMean of standardized features:")
print(X_wine_scaled_df.mean().round(6))

# Check standard deviation
print("\nStandard deviation of standardized features:")
print(X_wine_scaled_df.std().round(6))
```

```python
from sklearn.decomposition import PCA

# Create PCA with 2 principal components
pca_2 = PCA(n_components=2)

# Fit PCA and transform the standardized data
X_pca_2 = pca_2.fit_transform(X_wine_scaled)

# Convert transformed data to DataFrame
X_pca_2_df = pd.DataFrame(
    X_pca_2,
    columns=['PC1', 'PC2']
)

# Display first 5 transformed samples
print("First 5 rows after PCA:")
display(X_pca_2_df.head())

# Display shape of transformed dataset
print("\nOriginal dataset shape:", X_wine.shape)
print("Transformed dataset shape:", X_pca_2_df.shape)
```

```python
# Get explained variance ratio
explained_variance = pca_2.explained_variance_ratio_

print("Explained Variance Ratio:")
print(f"PC1: {explained_variance[0]:.4f}")
print(f"PC2: {explained_variance[1]:.4f}")

# Calculate total variance retained by the first two components
total_variance = explained_variance.sum()

print(f"\nTotal variance retained by PC1 and PC2: {total_variance:.4f}")
print(f"Percentage of variance retained: {total_variance * 100:.2f}%")
```

```python
# Apply PCA with all components
pca_full = PCA()

X_pca_full = pca_full.fit_transform(X_wine_scaled)

# Explained variance ratio of all components
explained_variance_all = pca_full.explained_variance_ratio_

# Calculate cumulative explained variance
cumulative_variance = explained_variance_all.cumsum()

# Create a table
variance_table = pd.DataFrame({
    'Principal Component': range(1, len(explained_variance_all) + 1),
    'Explained Variance Ratio': explained_variance_all,
    'Cumulative Variance': cumulative_variance
})

# Display the table
display(variance_table)

# Find minimum number of components required for 95% variance
n_components_95 = (
    cumulative_variance >= 0.95
).argmax() + 1

print(
    f"\nMinimum number of components required "
    f"to retain at least 95% variance: {n_components_95}"
)

print(
    f"Cumulative variance retained by these components: "
    f"{cumulative_variance[n_components_95 - 1] * 100:.2f}%"
)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

# Plot each wine class separately
for target_class in sorted(y_wine.unique()):
    plt.scatter(
        X_pca_2[y_wine == target_class, 0],
        X_pca_2[y_wine == target_class, 1],
        label=f"Class {target_class}",
        alpha=0.7
    )

plt.xlabel("Principal Component 1 (PC1)")
plt.ylabel("Principal Component 2 (PC2)")
plt.title("PCA: Wine Dataset Reduced to 2 Dimensions")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Create comparison table
comparison = pd.DataFrame({
    "Property": [
        "Number of samples",
        "Number of features",
        "Variance retained",
        "Dimensionality"
    ],
    "Original Dataset": [
        X_wine.shape[0],
        X_wine.shape[1],
        "100%",
        "13-dimensional"
    ],
    "PCA Dataset (2 Components)": [
        X_pca_2_df.shape[0],
        X_pca_2_df.shape[1],
        f"{total_variance * 100:.2f}%",
        "2-dimensional"
    ]
})

display(comparison)
```

```python
# Get PCA loadings for PC1 and PC2
loadings = pd.DataFrame(
    pca_2.components_.T,
    columns=['PC1', 'PC2'],
    index=X_wine.columns
)

print("PCA Loadings:")
display(loadings)

# Sort features according to their absolute contribution
print("\nTop features contributing to PC1:")
display(
    loadings['PC1']
    .abs()
    .sort_values(ascending=False)
    .head(5)
)

print("\nTop features contributing to PC2:")
display(
    loadings['PC2']
    .abs()
    .sort_values(ascending=False)
    .head(5)
)
```

```python
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis

# Create LDA with 2 components
lda = LinearDiscriminantAnalysis(n_components=2)

# Fit LDA using the standardized Wine data and class labels
X_lda = lda.fit_transform(X_wine_scaled, y_wine)

# Convert to DataFrame
X_lda_df = pd.DataFrame(
    X_lda,
    columns=['LD1', 'LD2']
)

# Display first 5 transformed samples
print("First 5 rows after LDA:")
display(X_lda_df.head())

# Display shapes
print("\nOriginal dataset shape:", X_wine.shape)
print("LDA transformed dataset shape:", X_lda_df.shape)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

# Plot each wine class separately
for target_class in sorted(y_wine.unique()):
    plt.scatter(
        X_lda[y_wine == target_class, 0],
        X_lda[y_wine == target_class, 1],
        label=f"Class {target_class}",
        alpha=0.7
    )

plt.xlabel("Linear Discriminant 1 (LD1)")
plt.ylabel("Linear Discriminant 2 (LD2)")
plt.title("LDA: Wine Dataset Reduced to 2 Dimensions")
plt.legend()
plt.grid(True)
plt.show()
```
